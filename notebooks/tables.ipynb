```python
import json
import pandas as pd

def jsons_to_df(json_paths, method_names, columns=None):
    rows = []
    
    for path, name in zip(json_paths, method_names):
        try:
            with open(path, "r") as f:
                d = json.load(f)
        except FileNotFoundError:
            print(f"File not found {name}: {path}")
            d = {col: None for col in columns} 
        d["method"] = name
        rows.append(d)
    
    df = pd.DataFrame(rows)
    
    if columns:
        df = df[["method"] + columns]
    
    display(df)
    # return df

# columns = ['roc_auc_mean', 'roc_auc_std', 'accuracy_mean', 'accuracy_std']
columns = ['accuracy_mean', 'accuracy_std', 'precision_mean', 'precision_std', 'recall_mean', 'recall_std', "specificity_mean", "specificity_std", "f1_score_mean", "f1_score_std", "roc_auc_mean", "roc_auc_std"]
columns_m_specificity = ['accuracy_mean', 'accuracy_std', 'precision_mean', 'precision_std', 'recall_mean', 'recall_std',  "f1_score_mean", "f1_score_std", "roc_auc_mean", "roc_auc_std"]
method_names = ['Radiomics Attention Pooling', 'Radiomics Statistical Pooling', '2D images Googlenet', 'RPNet3D'] #, "3D MTL Attention Siamese"
```

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# -------------------------------------------------------
# 1. Read one JSON result
# -------------------------------------------------------
def load_result(path, model, config):
    with open(path, "r") as f:
        d = json.load(f)

    d["model"] = model
    d["config"] = config
    return d


# -------------------------------------------------------
# 2. Paths
# -------------------------------------------------------

stat_root = (
    "/mnt/l/Basic/divi/jstoker/slicer_pdac/Marius/"
    "Reza Morphology/codes/Results/MorphScoreStatisticalPoolingMLP"
)

mil_root = (
    "/mnt/l/Basic/divi/jstoker/slicer_pdac/Marius/"
    "Reza Morphology/codes/Results/MorphScoreRadiomicsMIL"
)


# Configurations to include in the figure
configs = {
    "All": "shape_boundary_intensity_texture",

    "− Shape": "boundary_intensity_texture",
    "− Boundary": "shape_intensity_texture",
    "− Intensity": "shape_boundary_texture",
    "− Texture": "shape_boundary_intensity",

    "Shape only": "shape",
    "Boundary only": "boundary",
    "Intensity only": "intensity",
    "Texture only": "texture",
}


# -------------------------------------------------------
# 3. Load all results
# -------------------------------------------------------
rows = []

for display_name, folder in configs.items():

    stat_path = f"{stat_root}/{folder}/fivefold_aggregated_results.json"
    mil_path = f"{mil_root}/{folder}/fivefold_aggregated_results.json"

    # Apply your path changer if necessary
    stat_path = root_path_changer([stat_path], old_root, new_root)[0]
    mil_path = root_path_changer([mil_path], old_root, new_root)[0]

    try:
        rows.append(
            load_result(
                stat_path,
                model="RadPool",
                config=display_name
            )
        )
    except FileNotFoundError:
        print("Missing:", stat_path)

    try:
        rows.append(
            load_result(
                mil_path,
                model="RadMIL",
                config=display_name
            )
        )
    except FileNotFoundError:
        print("Missing:", mil_path)


df = pd.DataFrame(rows)

display(df)
```

```python
# -------------------------------------------------------
# 4. Publication-quality ablation figure
# -------------------------------------------------------

metrics = [
    ("accuracy", "Accuracy"),
    ("f1_score", "F1-score"),
    ("roc_auc", "AUROC"),
]

config_order = [
    "All",
    "− Shape",
    "− Boundary",
    "− Intensity",
    "− Texture",
    "Shape only",
    "Boundary only",
    "Intensity only",
    "Texture only",
]

models = ["RadPool", "RadMIL"]

# Use colors that remain distinguishable when printed
colors = {
    "RadPool": "#4C78A8",
    "RadMIL": "#F58518",
}

fig, axes = plt.subplots(
    1,
    len(metrics),
    figsize=(15, 4.8),
    sharex=True
)

x = np.arange(len(config_order))
bar_width = 0.36


for ax, (metric, metric_title) in zip(axes, metrics):

    for i, model in enumerate(models):

        means = []
        stds = []

        for config in config_order:

            subset = df[
                (df["model"] == model) &
                (df["config"] == config)
            ]

            if len(subset) == 0:
                means.append(np.nan)
                stds.append(np.nan)
            else:
                means.append(
                    subset[f"{metric}_mean"].iloc[0]
                )
                stds.append(
                    subset[f"{metric}_std"].iloc[0]
                )

        offset = (i - 0.5) * bar_width

        ax.bar(
            x + offset,
            means,
            width=bar_width,
            yerr=stds,
            capsize=3,
            label=model,
            color=colors[model],
            edgecolor="black",
            linewidth=0.5,
            error_kw={
                "elinewidth": 1,
                "capthick": 1,
            }
        )

    ax.set_title(
        metric_title,
        fontsize=12,
        fontweight="bold"
    )

    ax.set_xticks(x)

    ax.set_xticklabels(
        config_order,
        rotation=45,
        ha="right",
        fontsize=9
    )

    ax.set_ylim(0, 1.0)

    ax.grid(
        axis="y",
        linestyle="--",
        alpha=0.3
    )

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)


axes[0].set_ylabel("Performance", fontsize=11)

# One shared legend
handles, labels = axes[0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 1.04),
    ncol=2,
    frameon=False,
    fontsize=10
)

plt.tight_layout()

plt.savefig(
    "morphology_ablation_study.pdf",
    bbox_inches="tight",
    dpi=300
)

plt.savefig(
    "morphology_ablation_study.png",
    bbox_inches="tight",
    dpi=600
)

plt.show()
```
